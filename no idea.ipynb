```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('./Material/train_data.csv')
```

drop id

```python
df.drop(columns=['id'], inplace=True)
```

check for duplicates

```python
df.duplicated().sum()
```

drop duplicates

```python
df.drop_duplicates(inplace=True)
```

check for nulls

```python
df.isna().sum()
```

handle nulls

```python
df.fillna({'Age': round(df['Age'].median())}, inplace=True)
df.fillna({'Gender': df['Gender'].mode()[0]}, inplace=True)
df.fillna({'work_type': 'Unknown'}, inplace=True)
df.fillna({'smoking_status': 'Unknown'}, inplace=True)
```

check for garbage

```python
uq_gender = df['Gender'].unique()
uq_chest_pain_type = df['Chest pain type'].unique()
uq_fbs = df['FBS over 120'].unique()
uq_ekg = df['EKG results'].unique()
uq_exercise_angina = df['Exercise angina'].unique()
uq_slope_st = df['Slope of ST'].unique()
uq_number_vessels_fluro = df['Number of vessels fluro'].unique()
uq_thallium = df['Thallium'].unique()
uq_work_type = df['work_type'].unique()
uq_smoking_status = df['smoking_status'].unique()
uq_heart_disease = df['Heart Disease'].unique()

print("Unique values in Gender:", uq_gender)
print("Unique values in Chest pain type:", uq_chest_pain_type)
print("Unique values in FBS over 120:", uq_fbs)
print("Unique values in EKG results:", uq_ekg)
print("Unique values in Exercise angina:", uq_exercise_angina)
print("Unique values in Slope of ST:", uq_slope_st)
print("Unique values in Number of vessels fluro:", uq_number_vessels_fluro)
print("Unique values in Thallium:", uq_thallium)
print("Unique values in work_type:", uq_work_type)
print("Unique values in smoking_status:", uq_smoking_status)
print("Unique values in Heart Disease:", uq_heart_disease)
```
